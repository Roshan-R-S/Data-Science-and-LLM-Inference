```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/synthetic_credit_card_customer_behavior_dataset.csv")

print("Dataset shape:", df.shape)
```

```python
missing_summary = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_percentage": df.isnull().mean() * 100
})

missing_summary.sort_values(
    "missing_percentage",
    ascending=False
)
```

```python
duplicate_count = df.duplicated().sum()

print("Duplicate records:", duplicate_count)
```

```python
validation_summary = pd.DataFrame({
    "check": [
        "Age",
        "Payment Ratio",
        "Credit Utilization",
        "EMI Count",
        "Credit Score"
    ],
    "invalid_records": [
        len(df[(df["Age"] < 18) | (df["Age"] > 100)]),
        len(df[(df["Payment_Ratio"] < 0) | (df["Payment_Ratio"] > 1)]),
        len(df[(df["Credit_Utilization"] < 0) | (df["Credit_Utilization"] > 1)]),
        len(df[df["EMI_Count"] < 0]),
        len(df[(df["Credit_Score"] < 300) | (df["Credit_Score"] > 850)])
    ]
})

validation_summary
```

```python
constant_columns = [
    col for col in df.columns
    if df[col].nunique() <= 1
]

print("Constant columns:", constant_columns)
```
