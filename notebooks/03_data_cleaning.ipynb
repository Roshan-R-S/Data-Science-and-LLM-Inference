```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/raw/synthetic_credit_card_customer_behavior_dataset.csv")

print("Original shape:", df.shape)
```

```python
df_clean = df.copy()

print("Clean dataset shape:", df_clean.shape)
```

```python
df_clean = df_clean.drop_duplicates()

print("Shape after removing duplicates:", df_clean.shape)
```

```python
print("Total missing values:", df_clean.isnull().sum().sum())
```

```python
# Final validation after cleaning

final_validation = {
    "Rows": len(df_clean),
    "Columns": len(df_clean.columns),
    "Missing values": df_clean.isnull().sum().sum(),
    "Duplicate records": df_clean.duplicated().sum(),
    "Invalid Age": ((df_clean["Age"] < 18) | (df_clean["Age"] > 100)).sum(),
    "Invalid Payment Ratio": ((df_clean["Payment_Ratio"] < 0) | (df_clean["Payment_Ratio"] > 1)).sum(),
    "Invalid Credit Utilization": ((df_clean["Credit_Utilization"] < 0) | (df_clean["Credit_Utilization"] > 1)).sum(),
    "Invalid EMI Count": (df_clean["EMI_Count"] < 0).sum(),
    "Invalid Credit Score": ((df_clean["Credit_Score"] < 300) | (df_clean["Credit_Score"] > 850)).sum()
}

final_validation
```

```python
df_clean.to_csv(
    "../data/processed/cleaned_data.csv",
    index=False
)

print("Cleaned dataset saved successfully.")
```
